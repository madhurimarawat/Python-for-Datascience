# 🎵 Cleaning and Restructuring Songs JSON

## 🧪 Objective

This Python script is designed to **read**, **clean**, and **restructure** the data inside the `api/songs.json` file. The original data contains HTML content, extra whitespace, and inconsistent formatting that we aim to fix. The final cleaned data is saved in a new file called `cleaned_songs.json`.

---

## 🔍 What We'll Do

* Use **Python's `json`, `re`, and `html` libraries** to:

  * Read the JSON file
  * Strip HTML tags and decode HTML entities
  * Convert duration to integer values
* Print the **cleaned data preview**
* Display the **number of valid entries** (rows)
* Save the structured and clean data into a new JSON file for future use

```python
import json
import re
from html import unescape

# 📂 File paths
INPUT_FILE = 'api/songs.json'
OUTPUT_FILE = 'api/cleaned_songs.json'

# 🧹 Helper function to remove HTML tags and decode HTML entities
def clean_html(raw_html):
    clean_text = re.sub(r'<[^>]+>', '', raw_html)  # Remove HTML tags
    return unescape(clean_text).strip().strip('"')  # Decode entities and trim quotes/space

# 📥 Load original JSON data
with open(INPUT_FILE, 'r', encoding='utf-8') as file:
    data = json.load(file)

# 🧾 Extract and clean song entries
cleaned_songs = []
for song in data.get("table", {}).get("songs", {}).get("song", []):
    cleaned_song = {
        "date": song.get("date", "").strip(),
        "artist": clean_html(song.get("artist", "")),
        "title": clean_html(song.get("title", "")),
        "duration": int(song.get("duration", "0").strip()) if song.get("duration", "").strip().isdigit() else None
    }
    cleaned_songs.append(cleaned_song)

# 🖨️ Print cleaned data preview
print("\n📄 Cleaned Songs Data:\n")
print(json.dumps(cleaned_songs[:6], indent=2, ensure_ascii=False))

# 🔢 Print number of rows (songs)
print(f"\n✅ Total songs cleaned: {len(cleaned_songs)}")

# 💾 Save cleaned data to a new JSON file
with open(OUTPUT_FILE, 'w', encoding='utf-8') as outfile:
    json.dump(cleaned_songs, outfile, indent=2, ensure_ascii=False)

print(f"\n📁 Cleaned data saved to '{OUTPUT_FILE}'")
```
